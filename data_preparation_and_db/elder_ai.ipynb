# **Data Set 1: Fall Detection**

```python
import os
from PIL import Image

train_images_dir = '/content/drive/MyDrive/depi project/data archieved/Fall Detection(dataset1)/fall_dataset/images/train'
val_images_dir = '/content/drive/MyDrive/depi project/data archieved/Fall Detection(dataset1)/fall_dataset/images/val'

def check_dataset_health(img_dir):
    valid_images = 0
    corrupted_images = 0

    for file in os.listdir(img_dir):
        if file.lower().endswith(('.png', '.jpg', '.jpeg')):
            img_path = os.path.join(img_dir, file)
            try:
                with Image.open(img_path) as img:
                    img.verify()
                valid_images += 1
            except Exception:
                corrupted_images += 1
                os.remove(img_path)

    return valid_images, corrupted_images

valid_train, corr_train = check_dataset_health(train_images_dir)
valid_val, corr_val = check_dataset_health(val_images_dir)

print("Fall Detection Dataset Health Check Results:")
print(f"Train Images: Valid = {valid_train}, Corrupted Handled = {corr_train}")
print(f"Validation Images: Valid = {valid_val}, Corrupted Handled = {corr_val}")
```

# **Data Set 2: CASAS Smart Home**

```python
import os
import pandas as pd

casas_data_dir = '/content/drive/MyDrive/depi project/data archieved/CASAS Smart Home(dataset2)/data'

def clean_casas_dataset(directory):
    total_files = 0
    cleaned_files = 0
    skipped_files = 0
    error_files = 0

    print("Starting CASAS Smart Home dataset cleaning and health check...\n")

    for file in os.listdir(directory):
        if file.endswith('.csv'):
            total_files += 1
            file_path = os.path.join(directory, file)

            try:
                df = pd.read_csv(file_path, low_memory=False, on_bad_lines='skip')

                if df.empty:
                    print(f"[Skipped] File is empty: {file}")
                    skipped_files += 1
                    continue

                initial_rows = len(df)

                df.drop_duplicates(inplace=True)
                df.dropna(how='all', inplace=True)

                df.to_csv(file_path, index=False)
                cleaned_files += 1
                print(f"[Cleaned Successfully] {file} | Rows processed: {initial_rows} -> Cleaned rows: {len(df)}")

            except Exception as e:
                error_files += 1
                print(f"[Error] Could not process {file}: {e}")

    print("\n" + "="*50)
    print("CASAS Smart Home Cleaning Summary:")
    print(f"Total CSV files found: {total_files}")
    print(f"Successfully cleaned: {cleaned_files}")
    print(f"Skipped (Empty): {skipped_files}")
    print(f"Errors encountered: {error_files}")
    print("="*50)

clean_casas_dataset(casas_data_dir)
```

# **Data Set 3: Accelerometer**

```python
import os
import pandas as pd

accelerometer_dir = '/content/drive/MyDrive/depi project/data archieved/Accelerometer(dataset3)'

def clean_accelerometer_dataset(directory):
    total_files = 0
    cleaned_files = 0
    skipped_files = 0
    error_files = 0

    print("Starting Accelerometer dataset cleaning and health check...\n")

    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith(('.csv', '.txt')):
                total_files += 1
                file_path = os.path.join(root, file)

                try:
                    # قراءة الملف مع تخطي الأخطاء
                    df = pd.read_csv(file_path, low_memory=False, on_bad_lines='skip')

                    if df.empty:
                        print(f"[Skipped] File is empty: {file}")
                        skipped_files += 1
                        continue

                    initial_rows = len(df)

                    # إزالة التكرارات والصفوف الفارغة
                    df.drop_duplicates(inplace=True)
                    df.dropna(how='all', inplace=True)

                    # حفظ الملف بعد التنظيف
                    df.to_csv(file_path, index=False)
                    cleaned_files += 1
                    print(f"[Cleaned Successfully] {file} | Rows processed: {initial_rows} -> Cleaned rows: {len(df)}")

                except Exception as e:
                    error_files += 1
                    print(f"[Error] Could not process {file}: {e}")

    print("\n" + "="*50)
    print("Accelerometer Dataset Cleaning Summary:")
    print(f"Total files found: {total_files}")
    print(f"Successfully cleaned: {cleaned_files}")
    print(f"Skipped (Empty): {skipped_files}")
    print(f"Errors encountered: {error_files}")
    print("="*50)

clean_accelerometer_dataset(accelerometer_dir)
```
